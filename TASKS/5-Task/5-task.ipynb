```python
df['Source_Label'].fillna(df['Source_Label'].mode()[0],inplace=True)
```

```python
dummies=pd.get_dummies(df['Source_Label'],prefix='col',dtype=int)
```

```python
df=pd.concat([df.drop(columns='Source_Label'),dummies],axis=1)
```
