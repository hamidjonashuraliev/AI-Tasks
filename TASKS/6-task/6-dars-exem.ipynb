# ML coding


```python
df=pd.read_csv("water_potability.csv")
```

```python
df.info()
```

```python
#1. Mean orqali to’ldiring
df['Potability'].fillna(df["Potability"].mean(), inplace=True)
```

```python
# 2. Mode orqali to’ldiring
df["Potability"].fillna(df["Potability"].mode()[0], inplace=True)
```

```python
#3. Median orqali to’ldiring
df["Potability"].fillna(df["Potability"].median(), inplace=True)
```

```python
# 4. Fixed qiymat bn to’ldiring
df["Potability"].fillna('Y', inplace=True)
```

```python
# 5. Istalgan ustunni tashlab yuboring
df.drop("Turbidity", axis=1,inplace=True)
```
