```python
# Ixtiyoriy listda berilgan sonlarni yigindisini toping
# Ixtiyoriy list
numbers = [3, -2, 8, 22, 14]
yigindi=0
for son in numbers:
    yigindi +=son
print('sonlar yigindisi:', yigindi)
```

```python
# Ixtiyoriy listda berilgan sonlarni kopaytmasini toping
# Ixtiyoriy list
numbers = [3, 2, 7, 2, 4]
kopaytma=1
for son in numbers:
    kopaytma *=son
print('sonlar kopaytmasi:', kopaytma)
```

```python
# Ixtiyoriy listda berilgan just sonlarni yigindisini toping
# Ixtiyoriy list
numbers = [36, 12, 70, 24, 4]
yigindi=0
for son in numbers:
    if son % 2 == 0:
        yigindi += son
        print('juft sonlar yigindisi:', yigindi)
```

```python
# Ixtiyoriy listda berilgan toq sonlarni yigindisini toping
# Ixtiyoriy list
numbers = [33, 11, 7, 5, 41]
yigindi=0
for son in numbers:
    if son % 2 == 1:
        yigindi += son
        print('toq sonlar yigindisi:', yigindi)
```

```python
# Ixtiyoriy listda berilgan musbat va toq sonlarni yigindisini toping
# Ixtiyoriy list
numbers = [3, 11, 17, 13, 41]
yigindi=0
for son in numbers:
    if son % 2 == 1 and son > 0:
        yigindi += son
        print('musbat va toq sonlar yigindisi :', yigindi)
```

```python
# Ixtiyoriy listda berilgan musbat lekin toq bolmagan sonlarni yigindisini toping
# Ixtiyoriy list
numbers = [6, 12, 10, 18, 16]
yigindi=0
for son in numbers:
    if son % 2 == 0 and son > 0:
        yigindi += son
        print('musbat lekin toq bolmagan sonlarni :', yigindi)
```

```python
# Ixtiyoriy listda berilgan musbat va toq va 3 ga bolinadigan  sonlarni yigindisini toping
# Ixtiyoriy list
numbers = [3, 15, 9, 18, 21]
yigindi=0
for son in numbers:
    if son > 0 and son % 2 == 1 and son % 3 == 0 :
        yigindi += son
        print('musbat va toq va 3 ga bolinadigan sonlarni yigindisi:', yigindi)
```

```python
# Ixtiyoriy listda berilgan manfiy va 2  xonalari  sonlarni yigindisini toping
# Ixtiyoriy list
numbers = [-3, -15, -9, -18, -21]
yigindi=0
for son in numbers:
    if son < 0 and -99 < son < -9  :
        yigindi += son
        print('manfiy va 2  xonalari  sonlarni yigindisi :', yigindi)
```

```python
# Ixtiyoriy listda berilgan 5ga  karrali manfiy  sonlarni yigindisini toping
# Ixtiyoriy list
numbers = [-5, -15, -30, -10, -20]
yigindi=0
for son in numbers:
    if son % 5 ==0 and -99 < son < -9  :
        yigindi += son
        print('5ga  karrali manfiy  sonlarni yigindisi :', yigindi)
```

```python
# Ixtiyoriy listda berilgan 5ga  karrali manfiy  sonlarni kopaytmasini toping
# Ixtiyoriy list
numbers = [-5, -15, -30, -10, -20]
kopaytma=1
for son in numbers:
    if son % 5 == 0 and -99 < son < -9  :
        kopaytma *= son
        print('5ga  karrali manfiy  sonlarni kopaytmasi :', kopaytma)
```
