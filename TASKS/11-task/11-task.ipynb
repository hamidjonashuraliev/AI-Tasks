# Model predict


```python
import pandas as pd
```

```python
df=pd.read_csv('classification_df.csv')
```

```python
df.info()
```

# Data preprocessing


```python
for col in df.columns:
    if df[col].isnull().any():
        if df[col].dtype== 'object':
            df[col].fillna(df[col].mode()[0], inplace=True)
        else:
            df[col].fillna(df[col].mean(), inplace=True)
```

```python
df.info()
```

```python
df.head()
```

```python
df['rank-type'].value_counts()
```

```python
#Encoding
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
for col in df.columns.drop('rank-type'):
    if df[col].dtype=='object':
        if df[col].nunique()<=3:
            dummies=pd.get_dummies(df[col],prefix=col,dtype=int)
            df=pd.concat([df.drop(columns=col),dummies],axis=1)
        else:
            df[col]=encoder.fit_transform(df[col])
```

```python
df.info()
```

```python
df['rank-type']=encoder.fit_transform(df['rank-type'])
```

```python
df.info(

)
```

```python
df.nunique()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
num_col=df.select_dtypes(include=['int64', 'float64']).columns.drop('rank-type')
df[num_col]=scaler.fit_transform(df[num_col])
```

```python
df.head()
```

# Model Training


```python
x=df.drop('rank-type', axis=1)
```

```python
y=df['rank-type']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape, x_test.shape
```

```python

y_train.shape, y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
```

```python
model
```

```python
model.fit(x_train, y_train)
```

# Predicting


```python
y_pred=model.predict(x_test)
```

```python
y_pred[1]
```

# Evaluation


```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
score=accuracy_score(y_test,y_pred)
```

```python
score
```

```python
report=classification_report(y_test,y_pred)
```

```python
print(report)
```

# Linear Regression


```python
import pandas as pd
```

```python
df=pd.read_csv('WineQT.csv')
```

```python
df.info()
```

```python
df['pH']
```

```python
# Encoding
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
for col in df.columns:
    if df[col].dtype=='object':
        if df[col].nunique()<=3:
            dummies=pd.get_dummies(df[col],prefix=col,dtype=int)
            df=pd.concat([df.drop(columns=col),dummies],axis=1)
        else:
            df[col]=encoder.fit_transform(df[col])
```

```python
df.info()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
num_col=df.select_dtypes(include=['int64', 'float64']).columns.drop('pH')
df[num_col]=scaler.fit_transform(df[num_col])
```

```python
df.head()
```

```python
x=df.drop('pH', axis=1)
```

```python
y=df['pH']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score, mean_squared_error
```

```python
r2=r2_score(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
```
